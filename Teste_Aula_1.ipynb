```python
import random

def jogo_adivinhacao():
    """
    Jogo onde o usuário tenta adivinhar um número aleatório entre 1 e 100.
    """
    # Gera número aleatório entre 1 e 100
    numero_secreto = random.randint(1, 100)
    tentativas = 0

    print("=" * 50)
    print("🎯 BEM-VINDO AO JOGO DE ADIVINHAÇÃO! 🎯")
    print("=" * 50)
    print("Tente adivinhar o número secreto entre 1 e 100.")
    print("Digite 'sair' a qualquer momento para encerrar o jogo.")
    print("-" * 50)

    while True:
        try:
            # Recebe o palpite do usuário
            entrada = input("\nDigite seu palpite: ")

            # Verifica se o usuário quer sair
            if entrada.lower() == 'sair':
                print(f"\n😢 Que pena! O número secreto era {numero_secreto}.")
                print("Até a próxima!")
                break

            # Converte para número inteiro
            palpite = int(entrada)
            tentativas += 1

            # Valida se o número está no intervalo correto
            if palpite < 1 or palpite > 100:
                print("⚠️ Por favor, digite um número entre 1 e 100!")
                tentativas -= 1  # Não conta essa tentativa inválida
                continue

            # Verifica se o palpite está correto
            if palpite > numero_secreto:
                print("📈 Muito alto!")
            elif palpite < numero_secreto:
                print("📉 Muito baixo!")
            else:
                print("\n" + "=" * 50)
                print(f"🎉 PARABÉNS! 🎉")
                print(f"Você acertou o número {numero_secreto}!")
                print(f"Total de tentativas: {tentativas}")
                print("=" * 50)
                break

        except ValueError:
            print("❌ Entrada inválida! Por favor, digite um número inteiro entre 1 e 100 ou 'sair'.")

    print("\nObrigado por jogar! 🎮")

# Executa o jogo
if __name__ == "__main__":
    jogo_adivinhacao()
```

```python
# Programa simples para contar vogais e consoantes de uma palavra

palavra = input("Digite uma palavra: ")

# Define as vogais
vogais = "aeiouAEIOU"

# Inicializa contadores
cont_vogais = 0
cont_consoantes = 0

# Percorre cada letra da palavra
for letra in palavra:
    if letra.isalpha():  # Verifica se é uma letra
        if letra in vogais:
            cont_vogais += 1
        else:
            cont_consoantes += 1

# Exibe os resultados
print(f"Quantidade de vogais = {cont_vogais}")
print(f"Quantidade de consoantes = {cont_consoantes}")
```

```python
import metatrader as mt4
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
import numpy as np
import time

def configurar_conexao_mt4():
    """
    Configura e inicializa a conexão com o MetaTrader 4
    """
    print("🔄 Conectando ao MetaTrader 4...")

    # Inicializa a conexão com o MetaTrader 4
    # A biblioteca metatrader geralmente usa connect() em vez de initialize()
    try:
        if not mt4.initialize():
            print("❌ Erro ao inicializar MetaTrader 4")
            print("Verifique se o MetaTrader 4 está aberto")
            return False

        print("✅ Conexão estabelecida com sucesso!")

        # Tenta obter informações da conta
        try:
            account_info = mt4.account_info()
            if account_info:
                print(f"   Conta: {account_info.get('login', 'N/A')}")
                print(f"   Corretora: {account_info.get('server', 'N/A')}")
        except:
            print("   (Conta não logada ou informações indisponíveis)")

        return True

    except Exception as e:
        print(f"❌ Erro ao conectar: {e}")
        return False

def obter_dados_dolar_2026():
    """
    Obtém os dados históricos do Dólar para 2026 usando MT4
    """
    # Lista de possíveis símbolos para dólar brasileiro no MT4
    simbolos_possiveis = [
        "USDBRL",      # Formato comum
        "USD/BRL",     # Formato com barra
        "USD",         # Dólar americano
        "USDBRLm",     # Mini dólar
        "USDTRY",      # Alternativa
        "EURUSD"       # Fallback para teste
    ]

    simbolo = None

    # Tenta encontrar um símbolo válido
    print("\n🔍 Procurando símbolo do dólar...")
    for sym in simbolos_possiveis:
        try:
            # Verifica se o símbolo existe
            info = mt4.symbol_info(sym)
            if info:
                simbolo = sym
                print(f"✅ Símbolo encontrado: {simbolo}")
                break
        except:
            continue

    if simbolo is None:
        print("❌ Nenhum símbolo para dólar encontrado!")
        print("\nTentando obter lista de símbolos disponíveis...")
        try:
            symbols = mt4.symbols_get()
            if symbols:
                print("Símbolos disponíveis (primeiros 20):")
                for i, sym in enumerate(symbols[:20]):
                    print(f"  - {sym}")
                print("  ...")
        except:
            print("Não foi possível listar os símbolos")
        return None

    # Define as datas de 2026
    data_inicio = datetime(2026, 1, 1)
    data_fim = datetime(2026, 12, 31, 23, 59)

    print(f"\n📅 Coletando dados de {data_inicio.strftime('%d/%m/%Y')} a {data_fim.strftime('%d/%m/%Y')}")
    print(f"💱 Símbolo: {simbolo}")
    print("⏳ Aguarde enquanto os dados são baixados...\n")

    # No MT4, geralmente usamos copy_rates_from ou copy_rates_range
    # A API pode variar, vamos tentar diferentes abordagens
    rates = None

    try:
        # Tenta copiar os dados (formato pode variar)
        # Método 1: copy_rates_range (similar ao MT5)
        rates = mt4.copy_rates_range(simbolo, mt4.TIMEFRAME_D1, data_inicio, data_fim)

        if rates is None or len(rates) == 0:
            # Método 2: copy_rates_from com número de barras
            # Calcula o número de dias entre as datas
            dias = (data_fim - data_inicio).days
            rates = mt4.copy_rates_from(simbolo, mt4.TIMEFRAME_D1, data_fim, dias)

    except Exception as e:
        print(f"❌ Erro ao copiar dados: {e}")
        print("\nTentando método alternativo...")

        # Método alternativo: baixar em lotes
        try:
            rates = []
            data_atual = data_fim
            while data_atual > data_inicio:
                batch = mt4.copy_rates_from(simbolo, mt4.TIMEFRAME_D1, data_atual, 100)
                if batch is not None and len(batch) > 0:
                    rates.extend(batch)
                    data_atual = batch[0][0]  # Timestamp do primeiro dado
                else:
                    break
                time.sleep(0.1)  # Pequena pausa para não sobrecarregar
        except:
            pass

    if rates is None or len(rates) == 0:
        print(f"❌ Não foi possível obter dados para {simbolo}")
        print("\nPossíveis causas:")
        print("1. Dados históricos para 2026 não estão disponíveis")
        print("2. O símbolo não tem cotações para o período solicitado")
        print("3. Verifique as configurações de histórico no MT4")
        print("4. Pode ser necessário fazer download do histórico no MT4")
        return None

    # Converte para DataFrame do pandas
    try:
        # Formato típico do MT4: lista de tuplas ou dicionários
        if isinstance(rates[0], (tuple, list)):
            # Formato: (time, open, high, low, close, tick_volume, spread, real_volume)
            df = pd.DataFrame(rates, columns=['time', 'open', 'high', 'low', 'close', 'tick_volume', 'spread', 'real_volume'])
        elif isinstance(rates[0], dict):
            df = pd.DataFrame(rates)
        else:
            df = pd.DataFrame(rates)

        # Converte timestamp para datetime
        if 'time' in df.columns:
            df['time'] = pd.to_datetime(df['time'], unit='s')
            df.set_index('time', inplace=True)

        # Ordena por data
        df.sort_index(inplace=True)

    except Exception as e:
        print(f"❌ Erro ao converter dados: {e}")
        print("Estrutura dos dados recebidos:")
        print(rates[:2] if rates else "Vazio")
        return None

    print(f"✅ Dados obtidos com sucesso! Total de {len(df)} dias de cotação\n")
    if len(df) > 0:
        print(f"📅 Período: {df.index[0].strftime('%d/%m/%Y')} a {df.index[-1].strftime('%d/%m/%Y')}")
    return df

def calcular_estatisticas(df):
    """
    Calcula estatísticas da evolução do dólar
    """
    print("\n📊 ANALISANDO DADOS DO DÓLAR EM 2026")
    print("=" * 50)

    # Verifica se temos dados suficientes
    if len(df) == 0:
        print("❌ DataFrame vazio, não é possível calcular estatísticas")
        return None

    # Estatísticas básicas
    valor_inicial = df['close'].iloc[0]
    valor_final = df['close'].iloc[-1]
    maximo = df['close'].max()
    minimo = df['close'].min()
    data_maximo = df[df['close'] == maximo].index[0] if len(df[df['close'] == maximo]) > 0 else df.index[0]
    data_minimo = df[df['close'] == minimo].index[0] if len(df[df['close'] == minimo]) > 0 else df.index[0]

    # Variações
    variacao_abs = valor_final - valor_inicial
    variacao_percentual = (variacao_abs / valor_inicial) * 100 if valor_inicial != 0 else 0
    amplitude = maximo - minimo

    # Volatilidade (desvio padrão dos retornos diários)
    retornos_diarios = df['close'].pct_change().dropna()
    volatilidade = retornos_diarios.std() * 100 if len(retornos_diarios) > 0 else 0

    print(f"\n💰 VALORES DO DÓLAR EM 2026:")
    print(f"   • Valor inicial (primeiro dia): R$ {valor_inicial:.4f}")
    print(f"   • Valor final (último dia): R$ {valor_final:.4f}")
    print(f"   • Valor máximo: R$ {maximo:.4f} ({data_maximo.strftime('%d/%m/%Y')})")
    print(f"   • Valor mínimo: R$ {minimo:.4f} ({data_minimo.strftime('%d/%m/%Y')})")

    print(f"\n📈 EVOLUÇÃO:")
    print(f"   • Variação absoluta: R$ {variacao_abs:+.4f}")
    print(f"   • Variação percentual: {variacao_percentual:+.2f}%")
    print(f"   • Amplitude total: R$ {amplitude:.4f}")
    print(f"   • Volatilidade diária: {volatilidade:.2f}%")

    # Dados adicionais
    if 'open' in df.columns:
        dias_alta = (df['close'] > df['open']).sum()
        dias_baixa = (df['close'] < df['open']).sum()
        dias_estavel = (df['close'] == df['open']).sum()

        print(f"\n📆 DISTRIBUIÇÃO DOS PREGÕES:")
        print(f"   • Dias de alta: {dias_alta}")
        print(f"   • Dias de baixa: {dias_baixa}")
        print(f"   • Dias estáveis: {dias_estavel}")

    return {
        'inicial': valor_inicial,
        'final': valor_final,
        'maximo': maximo,
        'minimo': minimo,
        'variacao_percentual': variacao_percentual,
        'volatilidade': volatilidade
    }

def gerar_graficos(df):
    """
    Gera gráficos da evolução do dólar
    """
    print("\n🎨 GERANDO GRÁFICOS...")

    if len(df) == 0:
        print("❌ Sem dados para gerar gráficos")
        return

    # Configura o estilo dos gráficos
    try:
        plt.style.use('seaborn-v0_8-darkgrid')
    except:
        plt.style.use('default')

    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    fig.suptitle('Análise do Dólar em 2026 (MetaTrader 4)', fontsize=16, fontweight='bold')

    # 1. Gráfico de preços
    ax1 = axes[0, 0]
    ax1.plot(df.index, df['close'], 'b-', linewidth=2, label='Fechamento')
    ax1.fill_between(df.index, df['close'].min(), df['close'], alpha=0.3, color='blue')
    ax1.set_title('Evolução do Preço de Fechamento', fontsize=12, fontweight='bold')
    ax1.set_ylabel('Preço (R$)', fontsize=10)
    ax1.legend(loc='upper left')
    ax1.grid(True, alpha=0.3)

    # Adiciona média móvel
    if len(df) >= 20:
        media_20 = df['close'].rolling(window=20).mean()
        ax1.plot(df.index, media_20, 'r--', linewidth=1.5, label='MM 20 dias')
        ax1.legend(loc='upper left')

    # 2. Volume de negociação
    ax2 = axes[0, 1]
    if 'tick_volume' in df.columns:
        colors = ['g' if close >= open else 'r'
                  for close, open in zip(df['close'], df['open'])] if 'open' in df.columns else ['g'] * len(df)
        ax2.bar(df.index, df['tick_volume'], color=colors, alpha=0.7)
        ax2.set_title('Volume de Negociação', fontsize=12, fontweight='bold')
        ax2.set_ylabel('Volume', fontsize=10)
        ax2.set_xlabel('Data', fontsize=10)
        ax2.grid(True, alpha=0.3)

    # 3. Retornos diários
    ax3 = axes[1, 0]
    retornos = df['close'].pct_change() * 100
    ax3.bar(df.index, retornos, color='gray', alpha=0.7)
    ax3.axhline(y=0, color='black', linestyle='-', linewidth=1)
    if len(retornos) > 0:
        ax3.axhline(y=retornos.mean(), color='blue', linestyle='--',
                    linewidth=1, label=f'Média: {retornos.mean():.2f}%')
    ax3.set_title('Retornos Diários (%)', fontsize=12, fontweight='bold')
    ax3.set_ylabel('Retorno (%)', fontsize=10)
    ax3.set_xlabel('Data', fontsize=10)
    ax3.legend()
    ax3.grid(True, alpha=0.3)

    # 4. Distribuição de preços
    ax4 = axes[1, 1]
    ax4.hist(df['close'], bins=min(30, len(df)), color='skyblue', edgecolor='black', alpha=0.7)
    ax4.set_title('Distribuição dos Preços', fontsize=12, fontweight='bold')
    ax4.set_xlabel('Preço (R$)', fontsize=10)
    ax4.set_ylabel('Frequência', fontsize=10)
    ax4.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    print("✅ Gráficos gerados com sucesso!")

def main():
    """
    Função principal do programa
    """
    print("=" * 60)
    print("📊 ANÁLISE DA EVOLUÇÃO DO DÓLAR EM 2026")
    print("📌 Usando MetaTrader 4 (biblioteca metatrader)")
    print("=" * 60)
    print("\nEste programa irá:")
    print("• Conectar ao MetaTrader 4")
    print("• Baixar dados históricos do dólar para 2026")
    print("• Gerar estatísticas e gráficos da evolução")
    print("-" * 60)

    # Conecta ao MetaTrader 4
    if not configurar_conexao_mt4():
        print("\n⚠️ Não foi possível conectar ao MetaTrader 4.")
        print("Certifique-se de que:")
        print("1. O MetaTrader 4 está instalado")
        print("2. O MetaTrader 4 está aberto")
        print("3. A biblioteca metatrader está instalada: pip install metatrader")
        return

    try:
        # Obtém os dados do dólar
        df = obter_dados_dolar_2026()

        if df is None or len(df) == 0:
            print("\n⚠️ Não foi possível obter dados suficientes.")
            print("Possíveis causas:")
            print("1. MetaTrader 4 não está rodando")
            print("2. O símbolo do dólar não está disponível")
            print("3. Não há dados históricos para 2026")
            print("4. Verifique se você está logado no MT4")
            print("\nDica: No MT4, vá em Ferramentas -> Opções -> Gráficos")
            print("      e aumente o número máximo de barras no histórico")
            return

        # Calcula estatísticas
        stats = calcular_estatisticas(df)

        if stats is None:
            return

        # Gera gráficos
        gerar_graficos(df)

        print("\n" + "=" * 60)
        print("✅ ANÁLISE CONCLUÍDA!")
        print("=" * 60)

        # Resumo final
        print(f"\n📌 RESUMO FINAL:")
        print(f"   O dólar iniciou 2026 em R$ {stats['inicial']:.4f}")
        print(f"   e terminou em R$ {stats['final']:.4f}")
        print(f"   Variação total: {stats['variacao_percentual']:+.2f}%")
        print(f"   Volatilidade diária média: {stats['volatilidade']:.2f}%")

    except KeyboardInterrupt:
        print("\n\n⚠️ Programa interrompido pelo usuário.")
    except Exception as e:
        print(f"\n❌ Erro durante a execução: {e}")
        import traceback
        traceback.print_exc()

    finally:
        # Finaliza a conexão com o MetaTrader 4
        try:
            mt4.shutdown()
            print("\n🔌 Conexão com MetaTrader 4 encerrada.")
        except:
            pass
        print("👋 Programa finalizado!")

if __name__ == "__main__":
    main()
```

```python
def imprimir_triangulo(n):
    """
    Função que imprime um triângulo de números onde cada linha i
    contém o número i repetido i vezes.

    Args:
        n (int): Número de linhas a serem impressas
    """
    for i in range(1, n + 1):
        # Imprime o número i repetido i vezes
        for j in range(i):
            print(i, end="   ")
        print()  # Pula para a próxima linha


def main():
    """
    Função principal que solicita o valor n ao usuário e chama a função
    para imprimir o triângulo.
    """
    try:
        n = int(input("Digite um número inteiro positivo: "))

        if n <= 0:
            print("Por favor, digite um número positivo.")
        else:
            print(f"\nTriângulo para n = {n}:")
            print("-" * 30)
            imprimir_triangulo(n)

    except ValueError:
        print("Erro: Por favor, digite um número inteiro válido.")


# Executa o programa
if __name__ == "__main__":
    main()
```

```python
def soma_imposto(taxa_imposto, custo):
    """
    Função que adiciona o imposto sobre vendas ao custo do item.

    Args:
        taxa_imposto (float): Taxa de imposto em porcentagem (ex: 10 para 10%)
        custo (float): Custo do item antes do imposto

    Returns:
        float: Custo do item com o imposto incluído
    """
    # Calcula o valor do imposto
    valor_imposto = custo * (taxa_imposto / 100)

    # Adiciona o imposto ao custo
    custo_com_imposto = custo + valor_imposto

    return custo_com_imposto


def main():
    """
    Função principal que solicita os dados ao usuário e exibe o resultado.
    """
    print("=== Calculadora de Imposto sobre Vendas ===\n")

    try:
        # Solicita o custo do item
        custo = float(input("Digite o custo do item (R$): "))

        # Solicita a taxa de imposto
        taxa = float(input("Digite a taxa de imposto (%): "))

        # Chama a função para calcular o custo com imposto
        custo_final = soma_imposto(taxa, custo)

        # Exibe os resultados
        print("\n" + "=" * 40)
        print(f"Custo original: R$ {custo:.2f}")
        print(f"Taxa de imposto: {taxa:.1f}%")
        print(f"Valor do imposto: R$ {custo * (taxa / 100):.2f}")
        print("-" * 40)
        print(f"Custo com imposto: R$ {custo_final:.2f}")
        print("=" * 40)

    except ValueError:
        print("\nErro: Por favor, digite valores numéricos válidos!")


# Executa o programa principal
if __name__ == "__main__":
    main()
```
